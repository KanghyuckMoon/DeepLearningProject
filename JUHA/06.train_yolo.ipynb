# 니퍼를 제대로 인식하지 않는 문제가 발생하여. 새 데이터셋 폴더로 진행 예정.

```python
from pathlib import Path
import shutil
import random

project = Path("/home/user14/Documents/GitHub/DeepLearningProject")

old_dataset = project / "yolo_dataset"
new_dataset = project / "yolo_dataset_v2"
nipper_new = project / "nipper_one_box"

# 새 데이터셋 폴더 생성
for split in ["train", "val", "test"]:
    (new_dataset / "images" / split).mkdir(parents=True, exist_ok=True)
    (new_dataset / "labels" / split).mkdir(parents=True, exist_ok=True)

print("새 데이터셋 폴더 생성 완료")
```

2단계 — 기존 데이터 복사

```python
# 기존 데이터 복사
for split in ["train", "val", "test"]:
    src_images = old_dataset / "images" / split
    src_labels = old_dataset / "labels" / split

    dst_images = new_dataset / "images" / split
    dst_labels = new_dataset / "labels" / split

    for file in src_images.glob("*"):
        shutil.copy2(file, dst_images / file.name)

    for file in src_labels.glob("*.txt"):
        shutil.copy2(file, dst_labels / file.name)

print("기존 데이터 복사 완료")
```

3단계 — 새 니퍼 데이터 128장 추가

128장
↓
Train      90장
Validation 19장
Test       19장

```python
# 새 니퍼 이미지 목록
images = list((nipper_new / "images").glob("*"))

random.seed(42)
random.shuffle(images)

# 90 / 19 / 19로 분리
train_images = images[:90]
val_images = images[90:109]
test_images = images[109:]

print("새 니퍼 데이터")
print("Train :", len(train_images))
print("Val   :", len(val_images))
print("Test  :", len(test_images))
```

4단계 — 니퍼 데이터 복사

```python
def copy_nipper_data(image_list, split):

    for image_path in image_list:

        # 이미지 복사
        dst_image = new_dataset / "images" / split / image_path.name
        shutil.copy2(image_path, dst_image)

        # 같은 이름의 txt 찾기
        label_path = nipper_new / "labels" / f"{image_path.stem}.txt"

        if label_path.exists():
            dst_label = new_dataset / "labels" / split / label_path.name
            shutil.copy2(label_path, dst_label)


copy_nipper_data(train_images, "train")
copy_nipper_data(val_images, "val")
copy_nipper_data(test_images, "test")

print("새 니퍼 데이터 추가 완료")
```

5단계 — 데이터 개수 확인

```python
for split in ["train", "val", "test"]:

    image_count = len(list((new_dataset / "images" / split).glob("*")))
    label_count = len(list((new_dataset / "labels" / split).glob("*.txt")))

    print(f"{split}")
    print("  이미지:", image_count)
    print("  라벨  :", label_count)
```

6단계 — data.yaml 만들기

```python
yaml_text = f"""path: {new_dataset}
train: images/train
val: images/val
test: images/test

nc: 3

names:
  0: nipper
  1: pen
  2: wire_stripper
"""

(new_dataset / "data.yaml").write_text(yaml_text)

print(yaml_text)
```
