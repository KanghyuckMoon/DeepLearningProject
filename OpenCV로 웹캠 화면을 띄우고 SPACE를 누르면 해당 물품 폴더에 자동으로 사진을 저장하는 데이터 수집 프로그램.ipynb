## 2. 데이터 수집 프로그램

```python

import cv2
import os

# ==============================
# 1. 설정
# ==============================

# 촬영할 물품 이름
item_name = input("촬영할 물품 이름을 입력하세요: ")

# 데이터 저장 폴더
save_dir = os.path.join("dataset", item_name)

# 폴더가 없으면 자동 생성
os.makedirs(save_dir, exist_ok=True)

# 현재 폴더에 이미 저장된 사진 개수 확인
existing_files = [
    file for file in os.listdir(save_dir)
    if file.lower().endswith((".jpg", ".jpeg", ".png"))
]

image_count = len(existing_files)

print()
print("================================")
print("     물품 데이터 수집 프로그램")
print("================================")
print(f"물품 : {item_name}")
print(f"저장 위치 : {save_dir}")
print(f"현재 사진 수 : {image_count}장")
print()
print("[SPACE] 사진 촬영")
print("[Q]     프로그램 종료")
print("================================")


# ==============================
# 2. 웹캠 연결
# ==============================

camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("웹캠을 열 수 없습니다.")
    exit()


# ==============================
# 3. 카메라 화면
# ==============================

while True:

    # 카메라에서 한 프레임 읽기
    ret, frame = camera.read()

    if not ret:
        print("카메라에서 영상을 가져올 수 없습니다.")
        break

    # 화면에 안내 문구 표시
    display_frame = frame.copy()

    cv2.putText(
        display_frame,
        f"Item: {item_name}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    cv2.putText(
        display_frame,
        f"Images: {image_count}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    cv2.putText(
        display_frame,
        "SPACE: Capture / Q: Quit",
        (20, 120),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 255, 255),
        2
    )

    # 카메라 화면 출력
    cv2.imshow("Data Collector", display_frame)


    # ==============================
    # 4. 키보드 입력
    # ==============================

    key = cv2.waitKey(1) & 0xFF


    # SPACE → 사진 촬영
    if key == 32:

        filename = f"{item_name}_{image_count:04d}.jpg"

        filepath = os.path.join(
            save_dir,
            filename
        )

        cv2.imwrite(filepath, frame)

        image_count += 1

        print(f"저장 완료: {filepath}")


    # Q → 종료
    elif key == ord("q"):

        print("프로그램을 종료합니다.")
        break


# ==============================
# 5. 종료
# ==============================

camera.release()
cv2.destroyAllWindows()

print()
print("================================")
print("촬영 종료")
print(f"물품 : {item_name}")
print(f"총 사진 수 : {image_count}장")
print(f"저장 위치 : {save_dir}")
print("================================")


```
