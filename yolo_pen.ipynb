```python
from ultralytics import YOLO
import cv2

# 학습된 Pen 모델 불러오기
model = YOLO("runs/detect/train-11/weights/best.pt")

# 웹캠 연결
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("웹캠을 열 수 없습니다.")
else:
    print("웹캠 실행 중입니다. ESC를 누르면 종료됩니다.")

    while True:
        ret, frame = cap.read()

        if not ret:
            print("카메라 영상을 읽을 수 없습니다.")
            break

        # Pen 인식
        results = model(frame, conf=0.5, device="cpu")

        # 인식 결과 그리기
        annotated_frame = results[0].plot()

        # 화면 표시
        cv2.imshow("Pen Detection", annotated_frame)

        # ESC 키 → 종료
        if cv2.waitKey(1) & 0xFF == 27:
            break

    cap.release()
    cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2
from IPython.display import display, clear_output
from PIL import Image
import time

# 학습된 모델
model = YOLO("runs/detect/train-11/weights/best.pt")

# 카메라
cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

if not cap.isOpened():
    print("❌ 카메라를 열 수 없습니다.")
else:
    print("✅ 카메라 실행 중")

    try:
        while True:
            ret, frame = cap.read()

            if not ret:
                print("❌ 카메라 영상을 읽을 수 없습니다.")
                break

            # YOLO 추론 - CPU 사용
            results = model(
                frame,
                conf=0.5,
                device="cpu",
                verbose=False
            )

            # YOLO 결과 표시
            annotated_frame = results[0].plot()

            # BGR → RGB
            frame_rgb = cv2.cvtColor(
                annotated_frame,
                cv2.COLOR_BGR2RGB
            )

            # Notebook 화면 갱신
            clear_output(wait=True)
            display(Image.fromarray(frame_rgb))

            time.sleep(0.03)

    except KeyboardInterrupt:
        print("카메라 종료")

    finally:
        cap.release()
```

```python
from ultralytics import YOLO
import cv2
from IPython.display import display
from PIL import Image

# 1. 내가 학습한 Pen 모델 불러오기
model = YOLO("runs/detect/train-11/weights/best.pt")

# 2. 웹캠 연결
cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

if not cap.isOpened():
    print("❌ 카메라를 열 수 없습니다.")
else:
    print("✅ 카메라 연결 성공")
    print("펜을 카메라에 보여주세요.")

    # 3. 카메라에서 한 장 촬영
    ret, frame = cap.read()

    if ret:
        print("✅ 사진 촬영 성공")

        # 4. 학습한 Pen 모델로 인식
        results = model(
            frame,
            conf=0.25,
            device="cpu",
            verbose=False
        )

        # 5. 인식 결과 그리기
        result_frame = results[0].plot()

        # 6. 인식된 객체 확인
        boxes = results[0].boxes

        if len(boxes) > 0:
            print(f"🎯 {len(boxes)}개의 물체를 발견했습니다.")

            for box in boxes:
                confidence = float(box.conf[0])
                class_id = int(box.cls[0])
                class_name = model.names[class_id]

                print(
                    f"→ {class_name} "
                    f"(정확도: {confidence * 100:.1f}%)"
                )
        else:
            print("❌ 펜을 인식하지 못했습니다.")

        # 7. Notebook에 결과 표시
        result_frame = cv2.cvtColor(
            result_frame,
            cv2.COLOR_BGR2RGB
        )

        display(Image.fromarray(result_frame))

    else:
        print("❌ 카메라 영상을 읽을 수 없습니다.")

    cap.release()
```

```python
from ultralytics import YOLO
import cv2

# 학습한 모델 불러오기
model = YOLO("runs/detect/train-11/weights/best.pt")

# 학습된 클래스 확인
print("학습된 클래스:", model.names)

# 웹캠 연결
cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

if not cap.isOpened():
    print("❌ 카메라를 열 수 없습니다.")
else:
    print("✅ 카메라 실행 성공")
    print("물건을 카메라에 비춰보세요.")
    print("ESC 키를 누르면 종료됩니다.")

    while True:
        ret, frame = cap.read()

        if not ret:
            print("❌ 카메라 영상을 읽을 수 없습니다.")
            break

        # YOLO 실시간 추론
        results = model(
            frame,
            conf=0.25,
            device="cpu",
            verbose=False
        )

        # 탐지 결과를 카메라 화면에 표시
        annotated_frame = results[0].plot()

        # 카메라 창
        cv2.imshow(
            "YOLO Real-Time Detection",
            annotated_frame
        )

        # ESC 키를 누르면 종료
        if cv2.waitKey(1) & 0xFF == 27:
            break

    cap.release()
    cv2.destroyAllWindows()
```

```python
import torch
import torch.nn as nn

from torchvision import transforms
from PIL import Image

import cv2


# ==========================================
# 1. 클래스 이름
# ==========================================

classes = ['nipper', 'pen', 'wire_stripper']


# ==========================================
# 2. CNN 모델 구조
# ==========================================

class CNN(nn.Module):

    def __init__(self, num_classes):
        super().__init__()

        self.features = nn.Sequential(

            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(64 * 16 * 16, 128),
            nn.ReLU(),

            nn.Linear(128, num_classes)
        )

    def forward(self, x):

        x = self.features(x)
        x = self.classifier(x)

        return x


# ==========================================
# 3. 모델 생성
# ==========================================

model = CNN(len(classes))


# ==========================================
# 4. 학습한 모델 불러오기
# ==========================================

checkpoint = torch.load(
    "models/object_cnn.pth",
    map_location="cpu"
)

model.load_state_dict(checkpoint["model_state"])

model.eval()

print("✅ 학습 모델 불러오기 완료")
print("클래스:", classes)
```
